### Explication du résumé technique

* Combien y a-t-il d'entrées dans l'ensemble de données ?
Il y a 505 entrées (lignes) dans cet ensemble de données.

* Qu'est-ce que « non nul » ?
« Non nul » signifie que la cellule contient une donnée valide et n'est pas vide. Ici, toutes les colonnes ont 505 valeurs non nulles, ce qui indique que notre jeu de données est complet.

* Quels types de données sont présents dans l'ensemble de données ?
Il y a deux types de données présents : des nombres entiers et du texte.

* Quels sont ces types de données ?
- int64 : pour les nombres entiers (utilisé pour l'âge et les dépenses).
- str (ou object) : pour le texte (utilisé pour les autres colonnes comme le nom, le pays, etc.).

```python
print("--- Les Moyennes ---")
print("Moyenne Age :", df['Age'].mean())
print("Moyenne Dépenses :", df['Customer spendings'].mean())

print("\n--- Les Médianes ---")
print("Médiane Age :", df['Age'].median())
print("Médiane Dépenses :", df['Customer spendings'].median())

print("\n--- Bonus : Médiane Age par Pays ---")
print(df.groupby('Country')['Age'].median())
```

```python
import matplotlib.pyplot as plt

depenses_par_pays = df.groupby('Country')['Customer spendings'].sum()

plt.figure(figsize=(12, 6))
depenses_par_pays.plot(kind='bar', color='skyblue', edgecolor='black')

plt.title('Total des dépenses des clients par pays', fontsize=14, fontweight='bold')
plt.xlabel('Pays', fontsize=12)
plt.ylabel('Total des dépenses (€)', fontsize=12)

plt.xticks(rotation=90) 

plt.tight_layout()
plt.show()
```

```python
df_cleaned = df[df['Customer spendings'] >= 10]

df_cleaned = df_cleaned.drop_duplicates()

colonnes_a_garder = ['Country', 'Age', 'Gender', 'Customer spendings']
df_cleaned = df_cleaned[colonnes_a_garder]

df_cleaned.to_csv('dataset-sell4all-cleaned.csv', index=False)

print("Taille dyal les données 9bal l'nettoyage :", df.shape)
print("Taille dyal les données ba3d l'nettoyage :", df_cleaned.shape)
```
